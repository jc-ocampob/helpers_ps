```python
import numpy as np
import pandas as pd

import helpers_ps as hp


# Keep sparse MultiIndex rendering
pd.set_option(
    "styler.sparse.index",
    True,
)


# ---------------------------------------------------------------------
# Test DataFrame
# ---------------------------------------------------------------------

index = pd.MultiIndex.from_tuples(
    [
        ("Δ Yield", "MTD", "AAA"),
        ("Δ Yield", "MTD", "AA+"),
        ("Δ Yield", "MTD", "AA"),
        ("Δ Yield", "MTD", "AA-"),
        ("Δ Yield", "MTD", "A"),
        ("Δ Yield", "YTD", "AAA"),
        ("Δ Yield", "YTD", "AA+"),
        ("Δ Yield", "YTD", "AA"),
        ("Δ Yield", "YTD", "AA-"),
        ("Δ Yield", "YTD", "A"),
        ("Δ Spreads", "MTD", "AAA"),
        ("Δ Spreads", "MTD", "AA+"),
        ("Δ Spreads", "MTD", "AA"),
        ("Δ Spreads", "MTD", "AA-"),
        ("Δ Spreads", "MTD", "A"),
        ("Δ Spreads", "YTD", "AAA"),
        ("Δ Spreads", "YTD", "AA+"),
        ("Δ Spreads", "YTD", "AA"),
        ("Δ Spreads", "YTD", "AA-"),
        ("Δ Spreads", "YTD", "A"),
    ],
    names=[
        "Metric",
        "Period",
        "Rating",
    ],
)

columns = [
    "1M",
    "2M",
    "3M",
    "6M",
    "9M",
    "1Y",
    "3Y",
    "5Y",
    "7Y",
    "10Y",
]

rng = np.random.default_rng(10)

df_test = pd.DataFrame(
    rng.integers(
        low=-100,
        high=101,
        size=(len(index), len(columns)),
    ),
    index=index,
    columns=columns,
)


# ---------------------------------------------------------------------
# Color palette
# ---------------------------------------------------------------------

YIELD_COLOR = "#C23520"
SPREAD_COLOR = "#173B75"

SHORT_TERM_COLOR = "#3478DF"
MEDIUM_TERM_COLOR = "#B93622"
LONG_TERM_COLOR = "#26343A"

NEGATIVE_COLOR = "#FF0000"
POSITIVE_COLOR = "#008000"
ZERO_COLOR = "#B7B7B7"


# ---------------------------------------------------------------------
# Conditional text color
# ---------------------------------------------------------------------

def value_text_color(
    value: float,
) -> str:
    """
    Return a text color based on the sign of a value.

    Parameters
    ----------
    value : float
        Numeric value to evaluate.

    Returns
    -------
    str
        CSS text-color declaration.
    """
    if pd.isna(value):
        return ""

    if value == 0:
        return f"color: {ZERO_COLOR}"

    return hp.DfStyling.text_color_by_range(
        value=value,
        ranges={
            (-np.inf, 0): NEGATIVE_COLOR,
            (0, np.inf): POSITIVE_COLOR,
        },
    )


# ---------------------------------------------------------------------
# Apply general table settings
# ---------------------------------------------------------------------

styled_test = hp.DfStyling.style_table(
    obj=df_test,

    font_family="'Arial', sans-serif",
    font_size=10,
    text_align="right",

    # Remove Metric, Period, and Rating
    show_index_names=False,

    header_styles={
        "color": "white",
        "font-weight": "bold",
        "text-align": "center",
        "padding": "4px 8px",
    },

    index_styles={
        "color": "white",
        "font-weight": "bold",
        "text-align": "right",
        "padding": "3px 8px",
    },

    data_styles={
        "background-color": "white",
        "padding": "3px 8px",
        "text-align": "right",
    },

    column_header_styles=[
        {
            "columns": [
                "1M",
                "2M",
                "3M",
                "6M",
                "9M",
            ],
            "styles": {
                "background-color": SHORT_TERM_COLOR,
            },
        },
        {
            "columns": [
                "1Y",
                "3Y",
                "5Y",
            ],
            "styles": {
                "background-color": MEDIUM_TERM_COLOR,
            },
        },
        {
            "columns": [
                "7Y",
                "10Y",
            ],
            "styles": {
                "background-color": LONG_TERM_COLOR,
            },
        },
    ],

    row_index_styles=[
        # -------------------------------------------------------------
        # Δ Yield
        # -------------------------------------------------------------
        {
            "values": "Δ Yield",
            "level": 0,
            "styles": {
                "background-color": YIELD_COLOR,
                "color": "white",
                "font-weight": "bold",
                "text-align": "center",
            },
        },
        {
            "values": [
                "MTD",
                "YTD",
            ],
            "level": 1,
            "parent": ("Δ Yield",),
            "styles": {
                "background-color": YIELD_COLOR,
                "color": "white",
                "font-weight": "bold",
                "text-align": "center",
            },
        },
        {
            "values": [
                "AAA",
                "AA+",
                "AA",
                "AA-",
                "A",
            ],
            "level": 2,
            "parent": ("Δ Yield",),
            "styles": {
                "background-color": YIELD_COLOR,
                "color": "white",
                "font-weight": "bold",
                "text-align": "right",
            },
        },

        # -------------------------------------------------------------
        # Δ Spreads
        # -------------------------------------------------------------
        {
            "values": "Δ Spreads",
            "level": 0,
            "styles": {
                "background-color": SPREAD_COLOR,
                "color": "white",
                "font-weight": "bold",
                "text-align": "center",
            },
        },
        {
            "values": [
                "MTD",
                "YTD",
            ],
            "level": 1,
            "parent": ("Δ Spreads",),
            "styles": {
                "background-color": SPREAD_COLOR,
                "color": "white",
                "font-weight": "bold",
                "text-align": "center",
            },
        },
        {
            "values": [
                "AAA",
                "AA+",
                "AA",
                "AA-",
                "A",
            ],
            "level": 2,
            "parent": ("Δ Spreads",),
            "styles": {
                "background-color": SPREAD_COLOR,
                "color": "white",
                "font-weight": "bold",
                "text-align": "right",
            },
        },
    ],
)


# ---------------------------------------------------------------------
# Add section separators
# ---------------------------------------------------------------------

styled_test = hp.DfStyling.border_rows(
    obj=styled_test,
    rows=[
        ("Δ Yield", "MTD", "A"),
        ("Δ Yield", "YTD", "A"),
        ("Δ Spreads", "MTD", "A"),
        ("Δ Spreads", "YTD", "A"),
    ],
    position="bottom",
    styles={
        "width": "2px",
        "line_style": "solid",
        "color": "#000000",
    },
)


# ---------------------------------------------------------------------
# Format values
# ---------------------------------------------------------------------

styled_test = (
    styled_test
    .format("{:,.0f}")
    .map(value_text_color)
)


styled_test
```
