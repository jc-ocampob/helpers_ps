# Example: Using tag_dot within the graph_line function

`tag_dot` allows for inclusion of tags and dots based on the inherent data in the line chart, while preserving optionality of `GraphMtplt.tag()` and `GraphMtplt.dot`

## Imports

```python
import pandas as pd
import helpers_ps as hp
```

## Create Sample Data

```python
df = pd.DataFrame({
    "date": pd.date_range("2024-01-01", periods=12, freq="ME"),
    "return": [0.01, 0.02, -0.005, 0.015, 0.012, 0.018, -0.003, 0.009, 0.011, 0.014, 0.006, 0.013]
}).set_index("date")
```

## Create line graph with tags and dots

Example case for adding two sets of `tag_dot` with the following construction:

1. `group1`: Showing `tag` and `dot` for return on "2024-04-30" and last with template structure `"{y_value:,.2%}"` and dots color red and tag in bold.

2. `group2`: Showing `tag` and `dot` for return on "2024-01-31" and "2024-10-31" with template structure `"{x_value:%b-%Y}:\n{y_value:,.2%}"` and dots color blue.

```python
g = hp.GraphMtplt(dataframe=df)
g.graph_line(

    # Changing label for unique ticker 
    labels=["Demo1"],

    # Setting titles to the graph
    titles=dict(
        title="Demo Graph 1",
        subtitle="Brief description of the graph demo 1"
    ),

    # Setting the source to the graph 
    source= dict(
        text=["Source: helpers_ps"]
    ),

    # Setting Legend
    legend=dict(
        show=True,
        ncol=1,
    ),

    # Setting config of x-axis
    x_axis=dict(
        tick_step=3,
        bbg_format=True
    ),

    # Setting configo of y-axis
    y_axis=dict(
        fmt=",.2%",
        lim=(-0.01, 0.025)
    ),

    # Adding tags
    tag_dot={
        "1":dict(
            ticker="return",
            x_values=["last", pd.to_datetime("2024-04-30")],
            template="{y_value:,.2%}",
            legend_label="group1",
            show="dot_tag",
            dot=dict(
                color="red"
            ),
            tag=dict(
                bg_color="none",
                ubic_etq=(0,10),
                fontweight="bold"
            )
        ),
        "2":dict(
            ticker="return",
            x_values=[pd.to_datetime("2024-01-31"),pd.to_datetime("2024-10-31")],
            template="{x_value:%b-%Y}:\n{y_value:,.2%}",
            legend_label="group2",
            show="tag_dot",
            dot=dict(
                color="blue"
            )
        )
    }
)
```
