# Example: Creating a simple line graph

## Imports

```python
import pandas as pd
import helpers_ps as hp
```

## Create Sample Data

```python
df = pd.DataFrame({
    "date": pd.date_range("2024-01-01", periods=12, freq="ME"),
    "return": [0.01, 0.02, -0.005, 0.015, 0.012, 0.018, -0.003, 0.009, 0.011, 0.014, 0.006, 0.013]
}).set_index("date")
```

## Create simple line graph

Create a simple graph with the following in mind:

1. Relable `returns` to `Demo1`

2. Add `title` and `subtitle`

3. Add `Source`

4. Add `Legend`

5. Set `x_axis` with a tick step of 3 and  show the dates in Bloomberg style format

6. Set `y_axis` format to ,.2% and limit the ranges of the axis from -1% and 2.5%

```python
g = hp.GraphMtplt(dataframe=df)
g.graph_line(

    # Changing label for unique ticker 
    labels=["Demo1"],

    # Setting titles to the graph
    titles=dict(
        title="Demo Graph 1",
        subtitle="Brief description of the graph demo 1"
    ),

    # Setting the source to the graph 
    source= dict(
        text=["Source: helpers_ps"]
    ),

    # Setting Legend
    legend=dict(
        show=True,
        ncol=1,
    ),

    # Setting config of x-axis
    x_axis=dict(
        tick_step=3,
        bbg_format=True
    ),

    # Setting configo of y-axis
    y_axis=dict(
        fmt=",.2%",
        lim=(-0.01, 0.025)
    ),
)
```
