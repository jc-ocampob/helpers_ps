# Layout

```python
import helpers_ps as hp
import pandas as pd
```

Setting graph theme

```python
hp.set_graph_theme()
```

Creating a base dataframe

```python
df = pd.DataFrame({
    "date": pd.date_range("2024-01-01", periods=12, freq="ME"),
    "return": [0.01, 0.02, -0.005, 0.015, 0.012, 0.018, -0.003, 0.009, 0.011, 0.014, 0.006, 0.013],
    "nav": [101, 103, 101.5, 102, 104, 106, 105, 106.5, 107, 108, 109, 108.5]
}).set_index("date")

display(df)
```

## Adding Figure Level title

The [`.add_title()`](../api/GraphMtplt/layout#Title) method generates a figure level title and stores the information in the metadata.

```python
g = (
    hp.GraphMtplt(df)
    .plot()
    .add_title("Portfolio Performance")
)
```

## Adding Figure Level subtitle

The [`.add_subtitle()`](../api/GraphMtplt/layout.md#Subtitle) method generates a figure level subtitle and stores the information in the metadata.

```python
g = (
    hp.GraphMtplt(df)
    .plot()
    .add_title("Portfolio Performance")
    .add_subtitle("Histórical performance of net asset value of the portfolio")
)
```

## Adding Figure Level source

The [`.add_source()`](../api/GraphMtplt/layout.md#Source) method generates a figure level source and stores the information in the metadata.

```python
g = (
    hp.GraphMtplt(df)
    .plot()
    .add_title("Portfolio Performance")
    .add_subtitle("Histórical performance of net asset value of the portfolio")
    .add_source(["Source: Bloomberg", "(c)Helpers_ps"])
)
```

## Adding plot level title

The [`.add_plot_title()`](../api/GraphMtplt/layout.md#plot-level-title) method generates a figure level source and stores the information in the metadata.

```python
g = (
    hp.GraphMtplt(df)
    .plot(nrows=2, height_ratios=[0.2, 0.4], hspace=0.35)
    .subplot_adjust(top=0.8, bottom=0.2)
    .add_title("Portfolio Performance")
    .add_subtitle("Histórical performance of net asset value of the portfolio")
    .add_source(["Source: Bloomberg", "(c)Helpers_ps"])
    .add_plot_title("Net Asset Value")
    .axis(1)
    .add_plot_title("daily performance")
)
```
