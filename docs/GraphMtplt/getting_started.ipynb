## Figure level layout components

| Method | Description |
|---|---|
| [`add_title()`](layout.ipynb) | Adds figure level title |

## Plot level layout components

| Method | Description |
|---|---|
| `add_plot_title()` | Adds plot level title |
| `tag()` | Adds text tag to plot |
| `dot()` | Adds dot to plot |
| `shade_x()` | shades x to plot |
| `add_legend()` | Adds legend to figure |
| `add_legend_point()` | Adds dot to legend |
| `add_recessions()` | Adds recessions to time x-axis |
| `horizontal_guides()` | Adds horizontal guides to plot |
| `vertical_guides()` | Adds vertical guides to plot |
| `horizontal_lines()` | Adds horizontal lines to plot |
| `vertical_lines()` | Adds vertical lines to plot |
