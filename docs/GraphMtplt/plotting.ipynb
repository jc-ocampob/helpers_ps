# Ploting

```python
import helpers_ps as hp
import pandas as pd
```

Setting graph theme

```python
hp.set_graph_theme()
```

Creating a base dataframe

```python
df = pd.DataFrame({
    "date": pd.date_range("2024-01-01", periods=12, freq="ME"),
    "return": [0.01, 0.02, -0.005, 0.015, 0.012, 0.018, -0.003, 0.009, 0.011, 0.014, 0.006, 0.013],
    "nav": [101, 103, 101.5, 102, 104, 106, 105, 106.5, 107, 108, 109, 108.5]
}).set_index("date")

display(df)
```

## Creating a simple one plot figure

```python
g = (
    hp.GraphMtplt(df)
    .plot()
)
```

## Creating a simple multiplot figure

The [`.plot()`](../../api/GraphMtplt/plotting#plot) method generates a figure with customizable columns and rows as well as width heigh spaces and ratios 

```python
g = (
    hp.GraphMtplt(df)
    .plot(
        ncols=3,
        nrows=2,
        sharex=True,
        sharey=False,
    )
)
```

## Adjusting height and width ratios

```python
g = (
    hp.GraphMtplt(df)
    .plot(
        ncols=3,
        nrows=2,
        sharex=True,
        sharey=False,
        width_ratios=[0.2,0.3,0.5],
        height_ratios=[0.3,0.5]
    )
)
```

## Referencing subplot

SubPlot referencing is a array of the matrix representation where 0 is the defult selection and represents the left top. On the other hand n-1 references the right most bottom plot.

The [`.axis()`](../../api/GraphMtplt/plotting#axis-selection) method allows for plot selection

```python
g = (
    hp.GraphMtplt(df)
    .plot(
        ncols=3,
        nrows=2,
        sharex=True,
        sharey=False,
        width_ratios=[0.2,0.3,0.5],
        height_ratios=[0.3,0.7]
    )
    .axis(2)
    .graph_line(
        series=[{"ticker":"nav"}]
    )
)
```

## Adjusting subplot

The [`.subplot_adjust()`](../../api/GraphMtplt/plotting#subplot-adjust) method allows to re adjust the subplot of graph within the figure and inherits all subplot_adjust configuration directly linked to the figure 

```python
g = (
    hp.GraphMtplt(df)
    .plot(
        ncols=3,
        nrows=2,
        sharex=True,
        sharey=False,
        width_ratios=[0.2,0.3,0.5],
        height_ratios=[0.3,0.7]
    )
    .axis(2)
    .graph_line(
        series=[{"ticker":"nav"}]
    )
    .subplot_adjust(
        top=0.8,
        bottom=0.2,
        left=0.15,
        right=0.95
    )
)
```
