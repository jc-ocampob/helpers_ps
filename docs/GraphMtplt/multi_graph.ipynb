## Example of Multiplot

```python
import helpers_ps as hp
import pandas as pd
from matplotlib.colors import to_rgb
```

```python
hp.set_graph_theme()
df = pd.DataFrame({
    "date": pd.date_range("2024-01-01", periods=12, freq="ME"),
    "return": [0.01, 0.02, -0.005, 0.015, 0.012, 0.018, -0.003, 0.009, 0.011, 0.014, 0.006, 0.013],
    "nav": [101, 103, 101.5, 102, 104, 106, 105, 106.5, 107, 108, 109, 108.5]
}).set_index("date")
```

```python
graph = (
    hp.GraphMtplt(df)
    .plot(nrows=2)
    .graph_line(
        tickers="nav"
    )
    .axis(1)
    .graph_bar(
        tickers="return",
        colors="red",
        y_axis=dict(
            fmt=",.2%",
            fontsize="6"
        ),
        x_axis=dict(
            fontsize="7",
            fmt="%m-%Y"
        )
    )
    .add_titles(
        title="Valor Cuota historico",
        subtitle="dedjowieo"
    )
    .add_source(
        text="Helpers_ps 2026"
    )
)
```
