```python
import pandas as pd
from helpers_ps.MtpltGraph import GraphMtplt, set_graph_theme, buffers

# Optional: apply default chart style
set_graph_theme()

# Example dataframe
df = pd.DataFrame(
    {
        "Portfolio": [100, 102, 105, 103, 108],
        "Benchmark": [100, 101, 103, 104, 106],
    },
    index=pd.date_range("2024-01-31", periods=5, freq="ME")
)

# Create graph object
g = GraphMtplt(dataframe=df)

# Create line chart
g.graph_line(
    tickers=["Portfolio", "Benchmark"],
    labels=["Portfolio", "Benchmark"],
    titles={
        "title": "Portfolio vs Benchmark",
        "subtitle": "Indexed performance"
    },
    source={
        "source": "Source: Internal calculations"
    },
    legend={
        "show": True
    }
)

# Save to memory buffer
g.save(name="portfolio_vs_benchmark")

# Access image buffer
img = buffers["portfolio_vs_benchmark"]
```
