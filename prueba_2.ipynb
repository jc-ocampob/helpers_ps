```python
df = pd.DataFrame(
    {
        "Portfolio": [100, 102, 105, 103, 108],
        "Benchmark": [100, 101, 103, 104, 106],
    },
    index=pd.date_range("2024-01-31", periods=5, freq="ME")
)

hp.set_graph_theme()

g = (
    hp.GraphMtplt(dataframe=df)
    .plot(ncols=2, nrows=2, sharey=True, sharex=True)
    .graph_line(
        tickers="Portfolio",
        titles={
            "title": "Portfolio vs Benchmark"
        },
    )
    .add_source(["Fuente: Credicorp Capital"])
    .tag(pd.to_datetime("2024-03-31"), y_value=105, label="hol")
    .shade_x([("2023-12-31", "2024-03-31")])
    .axis(1)
    .graph_bar(
        tickers="Benchmark",
        axis_side="left",
    )
    .horizontal_guides()
)
```
