```python
import helpers_ps as hp
import pandas as pd
from helpers_ps.MtpltGraph.models import FigureTitle
```

```python
df = pd.DataFrame(
    {
        "Portfolio": [100, 102, 105, 103, 108],
        "Benchmark": [100, 101, 103, 104, 106],
    },
    index=pd.date_range("2024-01-31", periods=5, freq="ME")
)

hp.set_graph_theme()

g = hp.GraphMtplt(dataframe=df)

g = (
    g
    .plot(ncols=2, width_ratios=[0.3, 0.3], wspace=0.4)
    .add_title("Portafolio vs Benchmark")
    .add_subtitle("Rendimiento promedio ult.12 meses")
    .add_source(["Fuente: Credicorp Capital"])
    .graph_line(
        tickers="Portfolio",
    )
    .tag(pd.to_datetime("2024-03-31"), y_value=105, label="hol")
    .add_legend(show=True)
    .add_plot_title("Rendimiento de benchmark")
    .axis(1)
    .graph_line(
        tickers="Benchmark",
        x_axis=dict(
            bbg_format=True,
             tick_step=2
        )
    )
    .config_yaxis(
        label="NAV",
        label_color="red",
        spine_color="black",
        fmt=",.0%"
    )
    .add_plot_title(
        "Rendimiento del portafolio",
        loc="left",
        fontweight="normal",
        color="red",
        fontstyle="oblique",
        y=1.01
    )

    
    
)
```

```python
df2 = pd.DataFrame(
    {
        "Activo": ["Renta Fija", "Renta Variable", "Alternativos"],
        "Benchmark": [0.2, 0.5, 0.3],
    },
).set_index("Activo")

hp.set_graph_theme()

g = hp.GraphMtplt(dataframe=[df2, df])

g = (
    g
    .plot(ncols=2, wspace=0.4,)
    .add_title("Portafolio vs Benchmark")
    .add_subtitle("Rendimiento promedio ult.12 meses")
    .add_source(["Fuente: Credicorp Capital"])
    .graph_pie(
        tickers="Benchmark",
        startangle=180,
        donut_width=0.5,
    )
    .axis(1)
    .graph_box_whiskers(
        df_index=1,
        tickers="all",
    )
    .add_plot_title("Prueba")
)
```
